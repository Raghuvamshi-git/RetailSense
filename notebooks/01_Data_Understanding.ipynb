```python
import torch
import torch.nn as nn
from torch.utils.data import TensorDataset, DataLoader

lstm_features = [
    "lag_7",
    "lag_28",
    "rolling_mean_7",
    "rolling_std_7",
    "sell_price",
    "month",
    "quarter",
    "is_weekend",
    "is_festival",
    "snap"
]

train_values = train_data[lstm_features].values.astype("float32")
train_targets = train_data["sales"].values.astype("float32")

test_values = test_data[lstm_features].values.astype("float32")
test_targets = test_data["sales"].values.astype("float32")
```

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()

train_scaled = scaler.fit_transform(train_values)
test_scaled = scaler.transform(test_values)
```

```python
def create_sequences(data, targets, sequence_length=7):
    X = []
    y = []

    for i in range(sequence_length, len(data)):
        X.append(data[i-sequence_length:i])
        y.append(targets[i])

    return np.array(X), np.array(y)

X_lstm_train, y_lstm_train = create_sequences(
    train_scaled,
    train_targets,
    sequence_length=7
)

X_lstm_test, y_lstm_test = create_sequences(
    test_scaled,
    test_targets,
    sequence_length=7
)

print("Training shape:", X_lstm_train.shape)
print("Testing shape:", X_lstm_test.shape)
```

```python
X_lstm_train_tensor = torch.tensor(X_lstm_train, dtype=torch.float32)
y_lstm_train_tensor = torch.tensor(y_lstm_train, dtype=torch.float32)

X_lstm_test_tensor = torch.tensor(X_lstm_test, dtype=torch.float32)
y_lstm_test_tensor = torch.tensor(y_lstm_test, dtype=torch.float32)
```

```python
train_dataset = TensorDataset(
    X_lstm_train_tensor,
    y_lstm_train_tensor
)

train_loader = DataLoader(
    train_dataset,
    batch_size=64,
    shuffle=False
)
```

```python
class LSTMModel(nn.Module):
    def __init__(self, input_size):
        super().__init__()

        self.lstm = nn.LSTM(
            input_size=input_size,
            hidden_size=64,
            num_layers=2,
            batch_first=True,
            dropout=0.2
        )

        self.fc = nn.Linear(64, 1)

    def forward(self, x):
        output, _ = self.lstm(x)
        output = output[:, -1, :]
        output = self.fc(output)

        return output.squeeze()
```

```python
lstm_model = LSTMModel(
    input_size=len(lstm_features)
)

criterion = nn.MSELoss()

optimizer = torch.optim.Adam(
    lstm_model.parameters(),
    lr=0.001
)
```

```python
epochs = 10

for epoch in range(epochs):

    lstm_model.train()

    total_loss = 0

    for X_batch, y_batch in train_loader:

        optimizer.zero_grad()

        predictions = lstm_model(X_batch)

        loss = criterion(
            predictions,
            y_batch
        )

        loss.backward()

        optimizer.step()

        total_loss += loss.item()

    average_loss = total_loss / len(train_loader)

    print(
        f"Epoch {epoch + 1}/{epochs} - Loss: {average_loss:.4f}"
    )
```

```python
lstm_model.eval()

with torch.no_grad():
    lstm_pred = lstm_model(X_lstm_test_tensor).numpy()
```

```python
lstm_mae = mean_absolute_error(
    y_lstm_test,
    lstm_pred
)

lstm_rmse = np.sqrt(
    mean_squared_error(
        y_lstm_test,
        lstm_pred
    )
)

print("LSTM MAE:", lstm_mae)
print("LSTM RMSE:", lstm_rmse)
```

```python
comparison = pd.DataFrame({
    "Model": [
        "XGBoost",
        "LightGBM",
        "LSTM"
    ],
    "MAE": [
        mae,
        lgbm_mae,
        lstm_mae
    ],
    "RMSE": [
        rmse,
        lgbm_rmse,
        lstm_rmse
    ]
})

comparison
```

```python
lead_time_days = 7
service_level_z = 1.65
```

```python
decision_data = test_data.copy()

decision_data["predicted_demand"] = y_pred

decision_data["demand_std"] = decision_data["rolling_std_7"].fillna(0)
```

```python
decision_data["safety_stock"] = (
    service_level_z *
    decision_data["demand_std"] *
    np.sqrt(lead_time_days)
)
```

```python
decision_data["reorder_point"] = (
    decision_data["predicted_demand"] * lead_time_days
    + decision_data["safety_stock"]
)
```

```python
decision_data["current_inventory"] = (
    decision_data["predicted_demand"] * 1.2
)
```

```python
decision_data["stockout_risk"] = np.where(
    decision_data["current_inventory"] < decision_data["reorder_point"],
    "HIGH",
    "LOW"
)
```

```python
decision_data["recommended_order"] = np.maximum(
    decision_data["reorder_point"]
    - decision_data["current_inventory"],
    0
)
```

```python
decision_data[
    [
        "id",
        "date",
        "predicted_demand",
        "current_inventory",
        "safety_stock",
        "reorder_point",
        "recommended_order",
        "stockout_risk"
    ]
].head(20)
```

```python
risk_ratio = (
    decision_data["reorder_point"]
    - decision_data["current_inventory"]
) / decision_data["reorder_point"].replace(0, np.nan)

decision_data["risk_score"] = (
    risk_ratio.clip(lower=0) * 100
).fillna(0)
```

```python
decision_data["risk_level"] = pd.cut(
    decision_data["risk_score"],
    bins=[-np.inf, 10, 30, 60, np.inf],
    labels=["LOW", "MEDIUM", "HIGH", "CRITICAL"]
)
```

```python
decision_data["recommendation"] = np.select(
    [
        decision_data["risk_level"] == "CRITICAL",
        decision_data["risk_level"] == "HIGH",
        decision_data["risk_level"] == "MEDIUM"
    ],
    [
        "Urgent stock replenishment required",
        "Reorder stock soon",
        "Monitor inventory closely"
    ],
    default="Inventory level is adequate"
)
```

```python
def run_price_scenario(data, price_change):

    scenario = data.copy()

    scenario["scenario_price"] = (
        scenario["sell_price"] * (1 + price_change)
    )

    scenario["scenario_predicted_demand"] = (
        xgb_model.predict(
            scenario[features].assign(
                sell_price=scenario["scenario_price"]
            )
        )
    )

    scenario["current_revenue"] = (
        scenario["sell_price"] *
        scenario["predicted_demand"]
    )

    scenario["scenario_revenue"] = (
        scenario["scenario_price"] *
        scenario["scenario_predicted_demand"]
    )

    scenario["revenue_change"] = (
        scenario["scenario_revenue"]
        - scenario["current_revenue"]
    )

    return scenario
```

```python
price_down_10 = run_price_scenario(
    decision_data,
    -0.10
)
```

```python
price_up_10 = run_price_scenario(
    decision_data,
    0.10
)
```

```python
scenario_comparison = pd.DataFrame({
    "Current Demand": decision_data["predicted_demand"],
    "Price -10% Demand": price_down_10["scenario_predicted_demand"],
    "Price +10% Demand": price_up_10["scenario_predicted_demand"],
    "Current Revenue": decision_data["current_inventory"] * 0 + decision_data["sell_price"] * decision_data["predicted_demand"],
    "Price -10% Revenue": price_down_10["scenario_revenue"],
    "Price +10% Revenue": price_up_10["scenario_revenue"]
})

scenario_comparison.head(20)
```
